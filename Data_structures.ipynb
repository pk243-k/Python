**what is a collection or Data structure?**

--> collection contains both homogenous and hetrogenous data.

--> **Homogenous Data**: If a collection or data structure has same type  of elements or single data type elements.

--> **Hetrogenous Data**: If a collection or data structure has more thaan one type of elements or many data type elements.



```python
# How many types of collections or data sructures?
# There are 4 types
# 1.List
# 2.Tuple
# 3.Set
# 4.Dictionary
```

**List**
> What is list?

--> List is ordered, mutable and allows the duplicates.

--> List is declared by using --> []

```python
l=['apple','banana','apple','mango','apple']
print(len(l))
print(type(l))
```

Why list is ordered?

--> The items/elements in the list is having a sequence of index position numbers.

--> The index can be starts from 0.

--> By using the index position numbers we can access specific element

```python
l=['apple','pineapple','banana','apple','mango']
# Access the pineapple
print(l[1])
```

```python
#print the elements pineapple to mango?
print(l[1:5])
```

**List is mutable**

--> A collection is mutable or changable if it follows 3 properties.

1.We can replace or swap the items in the collection.

2.We add new items in the collection.

3.We can remove the items from the collection.

```python
l=['py','java','c','c++','.net']
#replace the 'py' with python
#To replace the items in the list use index number
l[0]='python'
print(l)
```

```python
#dir() is function that returns methods those are applicable on object.
l=['py','java','c','c++','.net']
dir(l)
```

```python
l.append(10)
print(l)
#Append():- by default adds the items at the end of list
```

```python
#2.Insert(index,item):- used to add the item at particular index
l.insert(2,'apple')
print(l)
```

```python
# Can we add multiple items in to list by using append (or) insert methods?
# NO
```

```python
#3.Extend():- ussed to add multiple items from other collections or data structures.
l_num=[1,2,3,4]
#add the list of numbers (l_num) into l
l.extend(l_num)
print(l)
```

```python
#Remove methods in list
#1.pop(index):-by default it removes last item from the list
l.pop()
print(l)
```

```python
#2.remove(item)
l.remove(10)
print(l)
```

```python
#3.clear()
l.clear()
print(l)
```

```python
# Add items in list --> append(),insert(),extend()
# Remove items in list --> pop(),remove(),clear()
```

**Set**

--> Set is defined as unordered, immutable and won't allows the duplicates

```python
s={'apple','banana','mango','apple','mango'}
print(s)
```

```python
# Set is immutable so we cannot replace but we can add and remove the items.
#1.add():-used to add a single item to the set
s= {'banana', 'mango', 'apple'}
s.add(10)
print(s)
```

```python
#2.Update():- is used to add multiple items from the other collection to set
s1 = {10,2,4,5,6}
s.update(s1)
print(s)
```

```python
#Remove methods in set
#1.pop():-is a bad practice because we don't know which item will remove
s.pop()
print(s)
```

```python
#2.remove(item_name):
s.remove(10)
print(s)
```

```python
#3.discard():
s.discard('mango')
print(s)
```

```python
#what is diff b/w remove and discard?
# if the item does not exist in set,remove() method rise an error where in discard() method if the item exist or not it won't rise an error
```

```python
# clear():
s.clear()
print(s)
```

```python
#Tuple is immutable but how can we add or replace or remove items from it
#By converting the tuple in to list
t=(1,2,3,4)
t_l=list(t)
print(t_l)
```

**Dictionary**

--> Dictionary is defined as unordered, mutable and won't allows the duplicates

```python
#Dict won't allows the duplicates:
d={
    'name':'MET',
    'address':'Hyd',
    'courses':['DA','DS','AI'],
    'address':'pune'
}
print(d)
print(len(d))
```

```python
# Dictionary is unordered : The items doesnot have index
# how we can access the values from dictionary?
# By using "key names"
print(d['name'])
```

```python
#Access the AI
print(d['courses'][2])
```

```python
# Dictionary is mutable means we can replace , add and remove the items
#Replace the address pune to hyderabad
d['address']='Hyderabad'
print(d)
```

```python
# Add new key value pair "duration : 3-6 months"
d['duration']="3-6 months"
d
```

```python
# Update used to add items from other dictionary
d1={'skill':['py','java','linux']}
d.update(d1)
d
```

```python
#Remove methods
#1.pop(key)
d.pop('skill')
d
```

```python
#2.popitem():- by default remove last key value pair
d.popitem()
d
```

```python
#3. clear(): returns empty dictionary
d.clear()
d
```
