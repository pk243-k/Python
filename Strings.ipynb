```python
#string manipulation examples
s="Hello World!"

#convert the string to uppercase
print(s.upper())

#convert the string to lowercase
print(s.lower())

#remove leading and trailing whitespaces from the string
print(s.strip())

#replace all occurences of 'o' with 'x' in the string
print(s.replace('o','x'))

#count the no of occurences of 'o' in the string
print(s.count('o'))
```

```python
#count the vowels in the string
s=input()
s=s.lower()
a=s.count('a')
e=s.count('e')
i=s.count('i')
o=s.count('o')
u=s.count('u')
print(f"no of vowels in the string:{a+e+i+o+u}")
```

```python
#Grade calculator
m=int(input("marks in maths:"))
s=int(input("marks in science:"))
e=int(input("marks in english:"))
total=m+e+s
avg=total/3
percentage =(total/300)*100
grade=''
if percentage>90:
  grade='A'
elif percentage>80 and percentage<=90:
  grade='B'
elif percentage>70 and percentage<=80:
  grade='c'
else:
  grade='pass'

print(f"Total marks:{total}\naverage marks:{avg}\ngrade:{grade} ")
```

```python
#palindrome checker
a=input()
if a[::-1]==a:
  print("palindrome")
else:
  print("not palindrome")
```

```python
#largest number of three numbers
a=input()
b,c,d=a.split(",")
x=int(b)
y=int(c)
z=int(d)
if x>y and x>z:
  print(x)
elif y>x and y>z:
  print(y)
else:
  print(z)
```

```python
#leap year checker
a=int(input())
if a%400==0 or a%4==0 and a%100!=0:
  print('leap')
else:
  print('not leap')
```
