collections or data structures  --> 4types
1. List --> [] --> is ordered, mutable and allows the duplicates
        
        > ordered means items in list has index
        > mutable means we can add, remove and replace
        > adding methods in list --> append(),insert(index,item),extend()
        > replace methods in list --> using the index --> list[index] = "item"
        > remove methods in list --> pop(),remove(),clear()

2. Tuple --> () --> is ordered, immutable and allows the duplicates
         
         >Ordered means items in tuple has index
         >tuple is immutable --> we cannot remove, add and replace items in tuple

3. SET --> set() --> is unordered, immutable and won't allows duplicates

        >set is unordered means no index for items
        >Set is immutable we cannot replace the items but we can add and remove the items
        >Adding methods in set --> add(),update()
        >remove methods in set --> pop(),discard(),remove(),clear()

4. dictionary --> {key:value} --> is unordered, mutable and won't allow duplicates
        
        >dict is unordered means no index for key:value pairs
        >dict is mutable means we can add, replace and remove the items
        >adding methods :- update(),  dict['new_key']="value"
        >replace method :- dict['existing_key'] = "value"
        >remove methods in dictionary :- pop(key_name),popitem(),clear()

**List comprehension**

```python
#what is list?
#list is odered, mutable and allows the duplicates
#what is list comprehension?
#list comprehension offers smaller line of code where a new list is created from existing list
```

```python
fruits = ['apple','banana','pineapple','kiwi','orange','mango']
#write a program to cheack element having letter "e" and print it in a new list
new = [] #store the elements having letter "e"
for i in fruits:
  if "e" in i:
    new.append(i)
print(new)
```

```python
#syntax:  [item for item in collection_name <space> condition]
new = [i for i in fruits if "e" in i]
print(new)
```

```python
l = [3,4,6,5,7,5,4,5,6,7,9,8,2,3]
#print even data in one list and odd data in another list
even_new = [j for j in l if j%2==0]
print(even_new)
```

```python
odd_new = [k for k in l if k%2==1]
print(odd_new)
```

```python
#step1 : create a class customer
class customer():
  #step3 create attributes
  Bank_name = "HFC Bank"
  IFSC = "HFC000678"
  BRANCH = "Ameerpet"
  CITY = "HYD"
  STATE = "TG"
  #step4 create object attributes
  #before creating the object attributes initate object attributes inside the class
  def __init__(self,Name,accno,balance,phone):
    self.Name = Name
    self.accno = accno
    self.balance = balance
    self.phone = phone

  #step5 create methods
  #create a method to welcome a user -- welcome wish
  def welcome_wish(self):
    return(f"Hello {self.Name}! welcome to HFC bank {self.BRANCH} {self.CITY}")
  #create a method to check the balance
  def check_balance(self):
    return(f"Your current account balance is {self.balance}/- rupees only")
  #create a method to deposit the amount
  def deposit(self,amount):
    self.balance = self.balance + amount
    return(f"Your account has been credited with {amount}/-. Your updated balance {self.balance}/-")
  #create a method to withdraw the amount
  def withdraw(self,amount):
    if amount<=self.balance:
       self.balance = self.balance - amount
       print(f"Your account has been debited with {amount}/-. Your updated balance {self.balance}/-")
    else:
       print("Insufficient funds")
#step2 :- create an object
c1 = customer("ram",327262353,10000,97848373712)
c2 = customer("jhon",237356323,15000,6789327363)
#ACCESS THE class attrinutes
# print(c1.IFSC)
# print(c2.BRANCH)
#access the object attributes
# print(c1.balance)
# print(c2.accno)
#Acess the method
# print(c1.welcome_wish())
# print(c2.welcome_wish())
print(c1.check_balance())
print(c1.deposit(16000))
c1.withdraw(30000)
```
