```python
import numpy as np
```

```python
a=np.array([[1,2],[3,4]])
print(a)
```

```python
#print even numbers from the data
print(np.where(a%2==0,a,-1))
```

```python
#print only even data
print(a[a%2==0])
```

```python
#compute the indexes where the condition is true
#np.argwhere(condition)
print(np.argwhere(a%2==0))
```

```python
b=np.array([1,2,5,6,4,34,56,23,12])
print(np.argwhere(b%2==0).flatten())
```

```python
#reshaping of an array
#generate a range of array having numbers from 1 to 9
#np.arange(start,end,step)
a=np.arange(1,10)
print(a)
```

```python
#convert or reshape the above array in to 3 rows and 3 columns
#np.reshape(rows,colms)
print(a.reshape(3,3))
print(a.flatten())
```

```python
#Add new axis to array's
#axis=0  -->rows
#axis=1  -->columns
a=np.arange(10)
print(a)
print(a.shape)
```

```python
c=a[np.newaxis,:]  #add a axis at the row level and push all the values to the columns
print(c)
print(c.shape)
```

```python
d=a[:,np.newaxis] #add a axis at the column level and push all the values to the rows
print(d)
print(d.shape)
```

```python
#concatenation of array's
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
#concat or the join the array a and b row wise or column wise
#np.concatenate((arr1,arr2),axis=0/1)
print(np.concatenate((a,b),axis=0))
print(np.concatenate((a,b),axis=1))
```

```python
#Hstack :-join the stack array's horizontally in sequence
d=np.hstack((a,b))
print(d)
```

```python
#vstack :-join the stack array's vertically in sequence
e=np.vstack((a,b))
print(e)
```

```python
#copying
a=np.array([1,2,3])
b=a.copy()
b[1]=4
print(a)
print(b)
#when we np.copy() if we make any changes in the new array it won't impact on the old array
```
