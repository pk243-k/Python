```python
# 1. start creating the variables by using alphabets
sonage = 29
print(sonage)
```

```python
# 2. Don't create variables with numbers
10 = 'hello'
print(10)
```

```python
# Smart ways of writing variable
# --> Assign multiple values to the multiple variables and print it in a separate line
a,b,c = 10,20,30
print(a,b,c,sep="\n")
```

```python
#Assign single value to the multiple variables and print it in a separate line
a=b=c=10
print(a,b,c,sep="\n")

# OR
a=10
b=a
c=b
print(a,b,c,sep="\n")
```
