Introduction to Data visualisation

```python
#what is the purpose of data visualisation?
# for better understand of the data we do Data visualisation.
#what are all the components do we have in a plot or graph? or define anatomy of a graph
#1.co-ordinated :- x co-ordinate and y co-ordinate
#2.labels :- xlabel and ylabel
#3.title
#4.grids
#5.ticks :-names on scale of x-axis and y-axis
#6.legends :- different categorized color visualisation
#7.figure :- overall graph is known as figure

#we use 2 important libraries to do the visualisation
#1.matplotlib
#2.seaborn

#what is the diff b/w matplotlib and seaborn?
#matplotlib provides basic visualisation where it won't ose stats in backend
#seaborn use to give satatical analysis
#in seaborn we have clear understanding of data when compares to matplotlib
#matplotlib is used to define the seaborn graph (like x_label,y_label,ticks,grids,title)
#piechart is present in matplotlib where it is not present in seaborn
```

**matplotlib**

```python
#import the matplotlib
import matplotlib.pyplot as plt
#by default matplotlib gives the lineplot
height=[120,150,110,160,170,150,180]
#draw a line plot to check the distribution of height of student
plt.plot(height)
plt.show()
```

```python
height=[120,150,110,160,170,150,180]
weight=[65,75,60,80,65,85,90]
#check the relation b/w hright and weight of students
#syntax:- plt.plot(x,y)
plt.plot(weight,height)
plt.show()
```

```python
#instead of line plot use "o" at each data point
plt.plot(weight,height,"o")
#format the line
#there are 4 formats
#1.solid line --> "-"
#2.dashed line --> "--"
#3.dash dot line --> "-."
#4.dotted line --> ":"
```

```python
height=[120,150,110,160,170,150,180]
weight=[65,75,60,80,65,85,90]
#syntax :- plt.plot(x,y,"marker_symbol:color_code")   #marker_symbol --> +,P,p,*,h,H,o,x,X  color_code: r,y,b,g
plt.plot(weight,height,"o:g")
plt.show()
```

```python
#change the marker color and i don't want any connected lines
#mfc = marker face/fill color, ms = marker size, mec = marker edge color
height=[120,150,110,160,170,150,180]
weight=[65,75,60,80,65,85,90]
plt.plot(weight,height,"o",mfc="#72f542",mec="#4248f5",ms=10)
#syntax_line plot :- plt.plot(x,y,marker,mfc=,mec=,ms=)
```

**scatter** **plot**

```python
#scatter plot is used to check the distribution of 2 numerical columns (2 continous data)
#syntax :- plt.scatter(x,y,color=,s=)
plt.scatter(weight,height,color="r",s=100)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
#load sample data from seaborn (tips)
df= sns.load_dataset("tips")
df.head()
```

```python
plt.plot(df["total_bill"],df["tip"])
plt.xlabel("total_bill")
plt.ylabel("tips")
plt.show()
```

```python
plt.scatter(df["total_bill"],df["tip"])
plt.xlabel("total_bill")
plt.ylabel("tips")
plt.show()
```

```python
#check the distribution total_bill --> histogram
plt.hist(df['total_bill'],bins=15)
plt.show()
```

```python
#find the relation b/w total bill and tip by using scatter plots in seaborn
#syntax :- sns.scatter(data=,x=,y=,hue=,style=,size=)
plt.figure(figsize=(7,5))
sns.scatterplot(data=df,x='total_bill',y='tip')
plt.show()
```

```python
#find the relation b/w total bill and tip by using scatter plots in seaborn according to gender
plt.figure(figsize=(7,5))
sns.scatterplot(data=df,x='total_bill',y='tip',hue='sex',style='day',size="size")
plt.show()
```

**heatmap** :- is used show the relation b/w numerical columns using colors.

Q) what is corelation?
is used to check the strength of straight line or linear relationship b/w numerical columns

corelation values can be taken from -1 to +1

if the corelation value is greater than 0.2 then there is a strong linear relation b/w target the feature variable

if the corelation values b/w -0.2 to +0.2 there is no relation b/w the columns or variables

if the corelation value is less than -0.2 then there is negative relation between variables



```python
import seaborn as sns
import matplotlib.pyplot as plt
df=sns.load_dataset("tips")
df.head()
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm")
plt.show()
```
